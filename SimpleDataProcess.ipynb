```python
import pandas as pd
def ObtainGeneralAggregateData(maxUser):
    df = pd.DataFrame(columns = ['userID','GameplayDur' ,'NumCoinsCollected', 'NumBombHit', 
                        'NumObstaclesHit','NumTreasureCollected',
                         'norNumCoinsCollected','norNumBombHit', 'norNumTreasureCollected',])
    for userID in range(maxUser):

        numCoinsCollected = chunk_database[(chunk_database['userID'] == userID) ]['numCoinsCollected'].sum()
        numBombHit = chunk_database[(chunk_database['userID'] == userID) ]['numBombHit'].sum()
        numObstaclesHit= chunk_database[(chunk_database['userID'] == userID) ]['numObstaclesHit'].sum()
        numTreasureCollected = chunk_database[(chunk_database['userID'] == userID) ]['numTreasureCollected'].sum()

        numCoinsHave = chunk_database[(chunk_database['userID'] == userID) ]['numCoinsHave'].sum()
        numBomb = chunk_database[(chunk_database['userID'] == userID) ]['numBomb'].sum()
        numTreasureHave = chunk_database[(chunk_database['userID'] == userID) ]['numTreasureHave'].sum()
        
        normNumCoinsCollected = numCoinsCollected/ numCoinsHave
        norNumBombHit = numBombHit/ numBomb
        norNumTreasureCollected = numTreasureCollected/ numTreasureHave

        GameplayDur = chunk_database[(chunk_database['userID'] == userID) & (chunk_database['chunkType'] == 11)]['endFrame'].astype(int).iloc[0]/30.0
       
        data = {
            "userID": [userID],
            'GameplayDur':[GameplayDur],
            "NumCoinsCollected": [numCoinsCollected],
            "NumBombHit": [numBombHit],
            "NumObstaclesHit": [numObstaclesHit],
            "NumTreasureCollected": [numTreasureCollected],

            'norNumCoinsCollected':[normNumCoinsCollected],
            'norNumBombHit':[norNumBombHit],
            'norNumTreasureCollected':[norNumTreasureCollected],
     
        }
        
        # Create a DataFrame from the data dictionary
        new_df = pd.DataFrame(data)
        
        # Concatenate the new data to the existing DataFrame
        df = pd.concat([df, new_df], ignore_index=True)
        
    return df


general_df = ObtainGeneralAggregateData(maxUser)
output_Path = 'C:/Users/z5308/Desktop/VRTestingProject/data/Reflex_data/simple_3joints/GeneralAggregateData.csv'
general_df.to_csv(output_Path, index=False) 
```

```python
import os
import re

def rename_files(path):
    files = os.listdir(path)
    pattern = r"player(\d+)(.*\.csv)$"  # Pattern to match "PlayerX.csv" or "PlayerX_chunk.csv"
    
    for file in files:
        match = re.match(pattern, file)
        if match:
            user_id = int(match.group(1))   # Extract current user ID (X)
            new_user_id = user_id + 21     # Increment to start from 12
            suffix = match.group(2)         # Get the rest of the file name
            
            # Create new filename with updated user ID
            new_file_name = f"player{new_user_id}{suffix}"
            
            # Rename the file
            #os.rename(os.path.join(path, file), os.path.join(path, new_file_name))
            print(f"Renamed '{file}' to '{new_file_name}'")


# Example usage
path = 'C:/Users/z5308/Desktop/VRTestingProject/data/Reflex_data/old/ylc_1111'
move_files(path)
```

```python
import os
import shutil

def move_files(source, target):
    files = os.listdir(source)
    pattern = r"player(\d+)_(.*\.csv)$"  # Pattern to match "PlayerX.csv" or "PlayerX_chunk.csv"
    
    for file in files:
        match = re.match(pattern, file)
        if match:
            user_id = int(match.group(1))   # Extract current user ID (X)
            new_user_id = user_id + 21     # Increment to start from 12
            suffix = match.group(2)         # Get the rest of the file name
            
            # Create new filename with updated user ID
            item_path = os.path.join(source, file)
            os.makedirs(target, exist_ok=True)
            # Rename the file
            shutil.move(item_path, target)
            print(f"move  '{file}' to '{target}'")

# Example usage
path = 'C:/Users/z5308/Desktop/VRTestingProject/data/Reflex_data/simple_3joints/'
target = 'C:/Users/z5308/Desktop/VRTestingProject/data/Reflex_data/simple_3joints/otherInfo/'
move_files(path,target)
```
